#### $\textbf{\color{Maroon}inplace parameter:}$ inplace is a boolean parameter used in many DataFrame and Series methods (like drop(), dropna(), fillna(), rename(), and sort_values()) methods.
#### It determines whether the operation modifies the original object directly or returns a new one. 
#### $\textbf{\color{Maroon}How it Works ?}$ inplace=False (Default): -- The method performs the operation, leaves your original DataFrame untouched, and returns a brand-new copy of the DataFrame with the changes applied.
#### $\textbf{\color{Maroon}inplace=True:}$ The method modifies your original DataFrame directly and returns None.
